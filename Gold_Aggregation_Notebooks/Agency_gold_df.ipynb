```python
# pipeline base params
table_name = "agency_dim_gold"
```

```python
from pyspark.sql import functions as F

# Load Silver Agency
df = spark.read.format("delta").load(
    "abfss://dev_Silver@onelake.dfs.fabric.microsoft.com/lh_nyc_payroll_silver.Lakehouse/Tables/agency_data_silver")
display(df)
```

```python
# Build Agency Dimension
dim_agency = df.select(
    "agencyid",
    "agencyname"
).dropDuplicates()

# Write to Gold Lakehouse
dim_agency.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable(table_name)

print("Gold dim_agency table created successfully")
```
