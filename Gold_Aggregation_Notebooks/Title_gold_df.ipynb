```python
# pipeline base params
table_name = "title_dim_gold"
```

```python
from pyspark.sql import functions as F

# Load Silver Title
df = spark.read.format("delta").load(
    "abfss://dev_Silver@onelake.dfs.fabric.microsoft.com/lh_nyc_payroll_silver.Lakehouse/Tables/title_data_silver")
display(df)
```

```python
# Build Title Dimension
dim_title = df.select(
    "titlecode",
    "titledescription"
).dropDuplicates()

# Write to Gold Lakehouse
dim_title.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable(table_name)

print("Gold dim_title table created successfully")
```
