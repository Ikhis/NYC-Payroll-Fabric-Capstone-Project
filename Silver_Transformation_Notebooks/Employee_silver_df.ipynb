```python
# pipeline base params
table_name = "employee_data_silver"
```

```python
from pyspark.sql import functions as F
import re
```

```python
df = spark.read.format("csv").option("header","true").load("abfss://dev_Bronze@onelake.dfs.fabric.microsoft.com/lh_nyc_payroll_bronze.Lakehouse/Files/raw/EmpMaster.csv")
# df now is a Spark DataFrame containing CSV data from "abfss://dev_Bronze@onelake.dfs.fabric.microsoft.com/lh_nyc_payroll_bronze.Lakehouse/Files/raw/EmpMaster.csv".
display(df)
```

```python
# Silver Transformation
# Standardize columns
def standardize_columns(df):
    for c in df.columns:
        new_c = c.lower().strip().replace(" ", "_")
        new_c = re.sub(r'[^a-z0-9_]', '', new_c)
        df = df.withColumnRenamed(c, new_c)
    return df

employee_df = standardize_columns(df)

# Remove duplicates & nulls
employee_df = employee_df.dropDuplicates(["employeeid"]) \
                     .filter(F.col("employeeid").isNotNull())
```

```python
# Write Silver
employee_df.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable(table_name)

print("Silver Employee completed")
```
